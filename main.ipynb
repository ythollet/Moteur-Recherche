```python
from Corpus.corpus import Corpus
from SearchEngine.class_search_engine import SearchEngine
```

```python
corpus = Corpus(
    nom = "corpus_finance"
)

search_engine = SearchEngine(
    in_corpus = corpus
)
```

```python
# 2.1 — Créer les objets graphiques simples.
import ipywidgets as widgets

# Titre de l'interface
titre = widgets.Label(value = "Moteur de recherche")

# Barre de recherche
mots_cles = widgets.Text(
    description = "Mots clés",
    placeholder = "Exemple : investment bank"
)

# Nombre de documents a extraire
nombre_articles = widgets.IntSlider(
    value = 5,
    min = 1,
    max = corpus.nb_docs,
    step = 1,
    description = "Nombre d’articles à extraire",
    style = {"description_width": "initial"},
    layout = widgets.Layout(width = "500px")
)
```

```python
# 2.2 à 2.5 — Assembler l'interface et lancer une recherche au clic.
from contextlib import redirect_stdout
from io import StringIO
from IPython.display import display, clear_output

sortie = widgets.Output()
bouton_rechercher = widgets.Button(
    description = "Rechercher",
    button_style = "primary"
)

def clique_bouton(bouton):
    """Lance la recherche avec les valeurs des widgets.

    :param bouton: Bouton ayant déclenché la recherche.
    """
    with sortie:
        clear_output(wait = True)
        requete = mots_cles.value.strip()
        if not requete:
            print("Saisissez au moins un mot-clé.")
            return

        bouton.disabled = True
        try:
            # Masque le tableau imprimé par le moteur ; la barre Jupyter
            # reste visible, car elle utilise un widget graphique.
            with redirect_stdout(StringIO()):
                resultats = search_engine.search(
                    in_text = requete,
                    in_nb_docs_to_return = nombre_articles.value
                )
            display(resultats)
        except Exception as erreur:
            clear_output(wait = True)
            print(f"La recherche a échoué : {erreur}")
        finally:
            bouton.disabled = False

bouton_rechercher.on_click(clique_bouton)

interface = widgets.VBox([
    titre,
    mots_cles,
    nombre_articles,
    widgets.HBox([bouton_rechercher]),
    sortie
])
display(interface)
```
